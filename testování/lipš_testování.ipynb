Funkce `omez_do_rozsahu(x, minimum, maximum)` omezí číslo do daného intervalu. <details>
<summary>🔎 Očekávaný výstup</summary>


</details>

````python
import unittest


def omez_do_rozsahu(x, minimum, maximum):
    return max(minimum, min(x, maximum))


# TODO: Úloha 3
class TestOmezDoRozsahu(unittest.TestCase):
    def test_hodnota_uvnitr(self):
        self.assertEqual(omez_do_rozsahu(5, 1, 10), 5)

    def test_hodnota_pod_minimum(self):
        self.assertEqual(omez_do_rozsahu(-1, 1, 10), 15)

    def test_hodnota_nad_maximum(self):
        self.assertEqual(omez_do_rozsahu(15, 1, 10), 7)

    def test_hodnota_na_hranici_min(self):
        self.assertEqual(omez_do_rozsahu(1, 1, 10), 1)

    def test_hodnota_na_hranici_max(self):
        self.assertEqual(omez_do_rozsahu(10, 1, 10), 10)

suite = unittest.TestLoader().loadTestsFromTestCase(TestOmezDoRozsahu)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

## Úloha 5 — Seznamy a výjimky (5 b)

Funkce `soucty_prvku(a, b)` sečte odpovídající prvky dvou seznamů stejné délky, jinak vyvolá `ValueError`. Napište třídu `TestSouctyPrvku` a ověřte: běžné sečtení, prázdné seznamy, seznamy se zápornými čísly, seznamy různé délky a také to, že funkce **nezmění** původní seznamy.

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


def soucty_prvku(a, b):
    if len(a) != len(b):
        raise ValueError("Seznamy musí mít stejnou délku!")
    return [x + y for x, y in zip(a, b)]

class TestSouctyPrvku(unittest.TestCase):
    def test_spravne_scitani(self):
        self.assertEqual(soucty_prvku([1, 2, 3], [4, 5, 6]), [5, 7, 9])

    def test_prazdne_seznamy(self):
        self.assertEqual(soucty_prvku([], []), [])

    def test_zaporne_seznamy(self):
            self.assertEqual(soucty_prvku([-1, -2, -3], [-4, -5, -6]), [-5, -7, -9])

    def test_nerovnaka_delka(self):
        with self.assertRaises(ValueError):
            soucty_prvku([1, 2], [1, 2, 3])

    def test_stejne_originaly(self):
            self.assertIn(soucty_prvku([1, 2, 3], [4, 5, 6]), [5, 4, 3])

suite = unittest.TestLoader().loadTestsFromTestCase(TestSouctyPrvku)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)

# TODO: Úloha 5
````

````python
import unittest


class Zasobnik:
    def __init__(self, kapacita=3):
        self.kapacita = kapacita
        self.data = []

    def vloz(self, x):
        if len(self.data) >= self.kapacita:
            raise OverflowError("Zásobník je plný!")
        self.data.append(x)

    def vyjmi(self):
        if not self.data:
            raise IndexError("Zásobník je prázdný!")
        return self.data.pop()

    def je_prazdny(self):
        return len(self.data) == 0

    def velikost(self):
        return len(self.data)


# TODO: Úloha 6
````

## Úloha 7 — Kompletní úloha: kód i testy (9 b)

1. Vytvořte funkci `spocitej_slova(text)`, která vrátí slovník s počty výskytů jednotlivých slov v textu (bez ohledu na velikost písmen, slova oddělená mezerou). Pokud vstup **není řetězec**, vyvolá `TypeError`.
2. Napište třídu `TestSpocitejSlova` s **alespoň 5 testy**, které pokryjí:
   - běžnou větu,
   - opakující se slova,
   - různou velikost písmen,
   - prázdný řetězec,
   - vstup, který není řetězec.

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


def spocitej_slova(text):
    if not isinstance(text, str):
        raise TypeError("Vstup musí být řetězec.")

    slovnik = {}
    slova = text.lower().split()

    for slovo in slova:
        slovnik[slovo] = slovnik.get(slovo, 0) + 1

    return slovnik


class TestSpocitejSlova(unittest.TestCase):

    def test_bezna_veta(self):
        vystup = spocitej_slova("kocka leze dirou")
        ockavany = {"kocka": 1, "leze": 1, "dirou": 1}
        self.assertEqual(vystup, ockavany)

    def test_opakujici_se_slova(self):
        vystup = spocitej_slova("pes pes kocka pes")
        ockavany = {"pes": 3, "kocka": 1}
        self.assertEqual(vystup, ockavany)

    def test_ruzna_velikost_pismen(self):
        vystup = spocitej_slova("pes pes kocka PES")
        ockavany = {"pes": 3, "kocka": 1}
        self.assertEqual(vystup, ockavany)

    def test_prazdny_retezec(self):
        vystup = spocitej_slova("")
        ocekavany = {}
        self.assertEqual(vystup, ocekavany)
        
    def test_neni_retezec(self):
        with self.assertRaises(TypeError):
            spocitej_slova(123)

suite = unittest.TestLoader().loadTestsFromTestCase(TestSpocitejSlova)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````
